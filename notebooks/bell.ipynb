````python
import qiskit
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram

print("Qiskit", qiskit.__version__)
````

## 1. Um qubit em superposição

Todo qubit começa em |0⟩. A porta **Hadamard** (`h`) leva esse estado a uma superposição: ao medir, saem 0 ou 1 com 50% de chance cada.

````python
qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.measure_all()
qc1.draw("mpl")
````

Agora vamos executar o circuito 1000 vezes no simulador (`shots=1000`) e contar os resultados.

````python
sampler = StatevectorSampler()

resultado = sampler.run([qc1], shots=1000).result()
contagens = resultado[0].data.meas.get_counts()
print(contagens)
plot_histogram(contagens)
````

## 2. O estado de Bell

Com dois qubits, aplicamos `h` no qubit 0 e depois uma **CNOT** (`cx`): ela inverte o qubit 1 sempre que o qubit 0 vale 1.

✏️ **Complete a linha marcada** e rode a célula.

````python
qc = QuantumCircuit(2)
qc.h(0)
# COMPLETE AQUI: aplique uma CNOT com controle no qubit 0 e alvo no qubit 1

qc.measure_all()
qc.draw("mpl")
````

<details>
<summary>Ver a resposta</summary>

```python
qc.cx(0, 1)
```
</details>

````python
resultado = sampler.run([qc], shots=1000).result()
contagens = resultado[0].data.meas.get_counts()
print(contagens)
plot_histogram(contagens)
````

**O que observar:** só aparecem `00` e `11`, cada um perto de 50%. Nunca `01` ou `10`.

Cada qubit sozinho parece aleatório, mas os dois sempre concordam. Essa é a assinatura do **emaranhamento**.
